# Computing Statistics with Pandas

Statistics help us understand and summarize data in meaningful ways. In this notebook, you'll learn how to compute various statistical measures using pandas, including summary statistics, grouped statistics, and value counts. You'll work with a fitness dataset to practice these essential statistical analysis techniques.

## Objectives

- Load the dataset into a pandas DataFrame
- Compute summary statistics (mean, median, std, percentiles)
- Group data by categories and compare statistics across groups
- Use value_counts to explore frequency distributions

```python
# Import pandas and numpy
# pandas provides data manipulation capabilities, while numpy offers numerical operations
import pandas as pd
import numpy as np
from pathlib import Path
pd.set_option('display.float_format', '{:.2f}'.format)

# Set the path to the csv file
# (synthetic dataset in the Demos folder)
data_path = Path('../starter/synthetic_data.csv')

# Load the data into a pandas DataFrame
# The read_csv() function reads a CSV file and creates a DataFrame
df = pd.read_csv(data_path)
```

### Count values in a categorical column
Use the `value_counts()` method to count the occurrences of each unique value in the `activity_level` column. This helps you understand the distribution of categorical data.

```python
# Count the occurrences of each activity_level
df['activity_level'].value_counts()
```

### Create a subset by selecting individual columns
You can create a subset of the DataFrame by selecting specific columns. Pass a list of column names in double brackets `[['col1', 'col2']]` to select multiple columns. For a single column, you can use `df['column_name']` or `df[['column_name']]` (double brackets return a DataFrame instead of a Series).

```python
# Create a subset with selected columns (e.g., activity_level, age_group, and weekly_workouts)
df_subset = df[['activity_level', 'age_group', 'weekly_workouts']]
df_subset.head()
```

### Select all numerical columns
Use the `select_dtypes()` method with `include=['number']` to create a subset containing only columns with numeric data types (e.g., int and float). This is useful when you need to perform calculations or statistical analysis on numeric data only.

```python
# Select only columns with numeric data types (int, float)
df_numerical = df.select_dtypes(include=['number'])
df_numerical.head()
```

### Compute summary statistics
Calculate mean, median, standard deviation, and selected percentiles for the numerical columns. These statistics provide a comprehensive overview of the central tendency and spread of your data.

```python
# Compute summary statistics using agg() to apply multiple functions at once
# mean: average value
# median: middle value (50th percentile)
# std: standard deviation (measure of spread)
summary = df[['avg_daily_steps','sleep_hours']].agg(['mean', 'median', 'std'])
print('Summary Statistics:')
print(summary)
```

### Compute grouped statistics by activity level
Use `groupby` to compare statistics across different activity levels. This allows you to understand how numerical variables differ between categories, such as how VO2 estimates or weekly workouts vary by activity level.

```python
# Group by activity_level and compute statistics for selected numerical columns
# groupby() splits the data into groups based on activity_level, then we apply statistical functions
cols = ['avg_daily_steps', 'sleep_hours']
grouped = df.groupby('activity_level')[cols].agg(['mean', 'median', 'std'])
print('Statistics grouped by activity level:')
print(grouped)

```

### Pivot table

The `pivot_table` method provides an alternative way to aggregate data by categories. It creates a summary table with one category as rows (the `index`) and the aggregated values of selected columns as the table body. Here we use it to compute mean values by activity level for the same numerical columns used in the groupby above.

```python
# Use pivot_table to compute means by activity level
cols = ['avg_daily_steps', 'sleep_hours']
print('\nMean values by activity level (using pivot_table):')
pivot_mean = df.pivot_table(index='activity_level', values=cols, aggfunc='mean')
print(pivot_mean)
```

### Value counts for categorical and discrete columns
Use `value_counts()` to see how often each unique value appears. This is useful for categorical columns and discrete numerical variables to understand the distribution of categories or levels in your data.

```python
# Identify categorical columns (object/category) for value_counts
# value_counts() returns the frequency of each unique value, sorted in descending order
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

print('Value counts for categorical columns:')
for col in cat_cols:
    print(f'\n{col}:')
    print(df[col].value_counts())

# value_counts on a discrete numerical column (e.g. weekly_workouts)
# For columns with few unique values, value_counts shows the distribution
print('\nValue counts for weekly_workouts (discrete numerical):')
print(df['weekly_workouts'].value_counts().sort_index())
```
