# Computing Statistics with Pandas

Statistics help us understand and summarize data in meaningful ways. In this notebook, you'll learn how to compute various statistical measures using pandas, including summary statistics, grouped statistics, and value counts. You'll work with a student performance dataset to practice these essential statistical analysis techniques.

## Objectives

- Load the dataset into a pandas DataFrame
- Compute summary statistics (mean, median, std, percentiles)
- Group data by categories and compare statistics across groups
- Use value_counts to explore frequency distributions

```python
# Import pandas
import pandas as pd
# Path handles file paths consistently across operating systems
from pathlib import Path
pd.set_option('display.float_format', '{:.2f}'.format)

# Set the path to the csv file
# (synthetic dataset in the Exercises folder)
data_path = Path('../starter/synthetic_data.csv')

# Load the data into a pandas DataFrame
# The read_csv() function reads a CSV file and creates a DataFrame
df = pd.read_csv(data_path)
```

### Count values in a categorical column
Use the `value_counts()` method to count the occurrences of each unique value in the `major` column. This helps you understand the distribution of categorical data.

```python
# Count the occurrences of each major
df['major'].value_counts()
```

### Create a subset by selecting individual columns
You can create a subset of the DataFrame by selecting specific columns. Pass a list of column names in double brackets `[['col1', 'col2']]` to select multiple columns. For a single column, you can use `df['column_name']` or `df[['column_name']]` (double brackets return a DataFrame instead of a Series).

```python
# Create a subset with selected columns (e.g., major, class_year, and study_hours)
df_subset = df[['major', 'class_year', 'study_hours']]
df_subset.head()
```

### Select all numerical columns
Use the `select_dtypes()` method with `include=['number']` to create a subset containing only columns with numeric data types (e.g., int and float). This is useful when you need to perform calculations or statistical analysis on numeric data only.

```python
# Select only columns with numeric data types (int, float)
df_numerical = df.select_dtypes(include=['number'])
df_numerical.head()
```

### Compute summary statistics
Calculate mean, median, standard deviation, and selected percentiles for the numerical columns. These statistics provide a comprehensive overview of the central tendency and spread of your data.

```python
# Compute summary statistics using agg() to apply multiple functions at once
# mean: average value
# median: middle value (50th percentile)
# std: standard deviation (measure of spread)
summary = df[['study_hours','sleep_hours']].agg(['mean', 'median', 'std'])
print('Summary Statistics:')
print(summary)
```

### Compute grouped statistics by major
Use `groupby` to compare statistics across different majors. This allows you to understand how numerical variables differ between categories, such as how exam scores or study hours vary by major.

### Pivot table

The `pivot_table` method provides an alternative way to aggregate data by categories. It creates a summary table with one category as rows (the `index`) and the aggregated values of selected columns as the table body. Here we use it to compute mean values by major for the same numerical columns used in the groupby above.

```python
# Group by major and compute statistics for selected numerical columns
# groupby() splits the data into groups based on major, then we apply statistical functions
cols = ['study_hours', 'sleep_hours']
grouped = df.groupby('major')[cols].agg(['mean', 'median', 'std'])
print('Statistics grouped by major:')
print(grouped)
```

```python
# Use pivot_table to compute means by major
print('\nMean values by major (using pivot_table):')
pivot_mean = df.pivot_table(index='major', values=cols, aggfunc='mean')
print(pivot_mean)
```

### Value counts for categorical and discrete columns
Use `value_counts()` to see how often each unique value appears. This is useful for categorical columns and discrete numerical variables to understand the distribution of categories or levels in your data.

```python
# Identify categorical columns (object/category) for value_counts
# value_counts() returns the frequency of each unique value, sorted in descending order
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

print('Value counts for categorical columns:')
for col in cat_cols:
    print(f'\n{col}:')
    print(df[col].value_counts())

# value_counts on a discrete numerical column (e.g. coffee_cups)
# For columns with few unique values, value_counts shows the distribution
print('\nValue counts for coffee_cups (discrete numerical):')
print(df['coffee_cups'].value_counts().sort_index())
```
