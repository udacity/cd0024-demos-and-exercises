# Visualizing Data with Pandas

Visualization is a powerful tool for exploring and understanding data. In this notebook, you'll learn how to create various types of plots using pandas and matplotlib to visualize a **synthetic Student Performance** dataset (major, class_year, study_hours, sleep_hours, class_attendance, coffee_cups, exam_score, letter_grade). You'll discover how different visualization techniques can reveal patterns, distributions, and relationships in your data.

## Objectives

- Load the dataset into pandas
- Create bar charts to visualize categorical data
- Create histograms to explore the distribution of numerical data
- Create box plots to compare distributions across categories

```python
# ============================================================
# Student exercise: Import libraries and load the dataset
#
# ============================================================
# Step: import pandas library as pd
import pandas as pd

# Path handles file paths consistently across operating systems
from pathlib import Path

# Step: set the path to the csv file
# (synthetic dataset in the Exercises folder)
data_path = Path('../starter/synthetic_data.csv')

# Step: load the data into a pandas DataFrame using read_csv()
df = pd.read_csv(data_path)
```

### Visualize major distribution
Create a bar chart to visualize the count of students by major. This will help you understand which majors are most common in the dataset.

```python
# ============================================================
# Student exercise: Bar chart of major distribution
#
# Instruction:
# Create a bar chart showing the count of students by major using matplotlib.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import matplotlib.pyplot as plt
import matplotlib.pyplot as plt

# Step: use value_counts() to count occurrences of each major, then plot as a bar chart
df['major'].value_counts().plot(kind='bar')

# Step: add title and axis labels
plt.title('Number of Students by Major')
plt.xlabel('Major')
plt.ylabel('Number of Students')

# Step: rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Step: display the plot
plt.show()
```

### Visualize class year distribution
Create a bar chart to visualize the count of students by class year. This will show you the distribution of students across different academic years.

```python
# ============================================================
# Student exercise: Bar chart of class year distribution
#
# Instruction:
# Create a bar chart showing the count of students by class_year using hvplot.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import hvplot.pandas
import hvplot.pandas

# Step: use value_counts() to count occurrences of each class year
counts = df['class_year'].value_counts()

# Step: convert the Series to a DataFrame with proper column names
counts_df = counts.reset_index()
counts_df.columns = ['class_year', 'count']

# Step: create an interactive bar chart using hvplot with title and axis labels
counts_df.hvplot.bar(x='class_year', y='count', 
                     title='Number of Students by Class Year',
                     xlabel='Class Year',
                     ylabel='Number of Students',
                     rot=45,
                     width=800, height=500)
```

### Visualize the distribution of study hours
Create a histogram to explore the distribution of study_hours. A histogram shows how many students fall into different ranges of study hours, helping you understand study patterns.

```python
# ============================================================
# Student exercise: Histogram of study hours
#
# Instruction:
# Create a histogram of the study_hours column using hvplot.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import hvplot.pandas
import hvplot.pandas

# Step: create an interactive histogram using hvplot.hist()
# The bins parameter controls the number of intervals in the histogram
df.hvplot.hist(y='study_hours', bins=50,
               title='Distribution of Study Hours',
               xlabel='Study Hours (per week)',
               ylabel='Frequency',
               width=800, height=500)
```

### Visualize the distribution of exam scores
Create a histogram to explore the distribution of exam_score, which represents student performance on exams. This helps you understand the distribution of academic performance.

```python
# ============================================================
# Student exercise: Box plot comparing exam scores by major
#
# Instruction:
# Create a box plot comparing exam_score across different majors using hvplot.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import hvplot.pandas
import hvplot.pandas

# Step: create an interactive box plot using hvplot.box()
# The 'by' parameter groups the data by major
plot = df.hvplot.box(y='exam_score', by='major',
                     title='Exam Scores by Major',
                     xlabel='Major',
                     ylabel='Exam Score',
                     width=800, height=500)

# Rotate x-axis labels 45 degrees using Bokeh options
plot.opts(xrotation=45)
```

### Compare exam scores by major
Create a box plot to compare the distribution of exam scores across different majors. Box plots are excellent for comparing distributions across categories as they show median, quartiles, and outliers.

```python
# ============================================================
# Student exercise: Histogram of exam scores
#
# Instruction:
# Create a histogram of the exam_score column using hvplot.
# This shows the distribution of student exam performance.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import hvplot.pandas
import hvplot.pandas

# Step: create an interactive histogram using hvplot.hist()
# The bins parameter controls the number of intervals in the histogram
df.hvplot.hist(y='exam_score', bins=50,
               title='Distribution of Exam Scores',
               xlabel='Exam Score',
               ylabel='Frequency',
               width=800, height=500)
```

### Compare exam scores by class year
Create a box plot to compare the distribution of exam scores across different class years. This visualization helps identify patterns in academic performance across different academic levels.

```python
# ============================================================
# Student exercise: Box plot comparing exam scores by class year
#
# Instruction:
# Create a box plot comparing exam_score across different class years using hvplot.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import hvplot.pandas
import hvplot.pandas

# Step: create an interactive box plot using hvplot.box()
# The 'by' parameter groups the data by class_year
plot = df.hvplot.box(y='exam_score', by='class_year',
                     title='Exam Scores by Class Year',
                     xlabel='Class Year',
                     ylabel='Exam Score',
                     width=800, height=500)

# Rotate x-axis labels 45 degrees using Bokeh options
plot.opts(xrotation=45)
```

### Compare average exam scores across majors using a line plot
Create a line plot to visualize how the average exam score varies across different majors. Line plots are useful for showing trends and comparing values across categories in a continuous manner.

```python
# ============================================================
# Student exercise: Line plot of average exam scores by major
#
# Instruction:
# Create a line plot comparing average exam_score across majors using matplotlib.
# Be sure to add a title and label the axes.
# ============================================================
# Step: calculate the mean exam_score for each major and sort the values
avg_scores = df.groupby('major')['exam_score'].mean().sort_values()

# Step: create a line plot using plot(kind='line') with markers
avg_scores.plot(kind='line', marker='o')

# Step: add title and axis labels
plt.title('Average Exam Score by Major')
plt.xlabel('Major')
plt.ylabel('Average Exam Score')

# Step: add grid for better readability
plt.grid(True, alpha=0.3)

# Step: display the plot
plt.show()
```

### Explore relationships between numerical variables using a scatter plot
Create a scatter plot to visualize the relationship between two numerical variables. Scatter plots are excellent for identifying correlations, trends, and outliers in the data. This helps you understand how different features relate to each other, such as how study hours relate to exam scores.

```python
# ============================================================
# Student exercise: Scatter plot of study hours vs exam scores
#
# Instruction:
# Create a scatter plot comparing study_hours and exam_score using hvplot.
# Scatter plots help identify relationships and patterns between two numerical variables.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import hvplot.pandas
import hvplot.pandas

# Step: prepare the data by selecting relevant columns and removing missing values
subset = df[['study_hours', 'exam_score']].dropna()

# Step: create an interactive scatter plot using hvplot with title and axis labels
subset.hvplot.scatter(x='study_hours', y='exam_score',
                     title='Study Hours vs Exam Scores',
                     xlabel='Study Hours (per week)',
                     ylabel='Exam Score',
                     alpha=0.5,
                     width=800, height=500)
```

### Visualize correlations between numerical variables using a heatmap
Create a correlation heatmap to visualize the strength and direction of relationships between numerical variables. Correlation values range from -1 to 1, where values close to 1 indicate strong positive correlation, values close to -1 indicate strong negative correlation, and values near 0 indicate weak or no correlation.

```python
# ============================================================
# Student exercise: Correlation heatmap of numerical variables
#
# Instruction:
# Create a correlation heatmap for numerical features using pandas.
# Be sure to add a title and label the axes.
# ============================================================
# Step: identify numerical columns for correlation analysis
cols = ['study_hours', 'sleep_hours', 'class_attendance', 'coffee_cups', 'exam_score']

# Step: compute the correlation matrix
corr = df[cols].corr()

# Step: create the heatmap using pandas style.background_gradient()
# cmap='coolwarm' uses a color scheme, vmin and vmax set the color scale range
# The background_gradient method applies a color gradient to the DataFrame cells
corr.style.background_gradient(cmap='coolwarm', vmin=-1, vmax=1)
```

### Visualize top majors using a horizontal bar chart
Create a horizontal bar chart to compare the count of students across different majors. Horizontal bar charts are useful when you have many categories or long category names, as they provide more space for labels.

```python
# ============================================================
# Student exercise: Horizontal bar chart of major distribution
#
# Instruction:
# Create a horizontal bar chart showing the count of students by major using matplotlib.
# Horizontal bar charts are useful when category names are long or when you want to emphasize comparisons.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import matplotlib.pyplot as plt
import matplotlib.pyplot as plt

# Step: use value_counts() to count occurrences of each major
counts = df['major'].value_counts()

# Step: create the horizontal bar chart using matplotlib
plt.barh(counts.index, counts.values)

# Step: add title and axis labels
plt.xlabel('Number of Students')
plt.ylabel('Major')
plt.title('Number of Students by Major (Horizontal)')

# Step: display the plot
plt.show()
```

### Explore relationships between three variables using a 3D scatter plot
Create a 3D scatter plot to visualize relationships between three numerical variables simultaneously. This allows you to understand complex relationships that might not be apparent in 2D plots. Here we explore how study hours, exam scores, and sleep hours relate to each other, colored by major.

```python
# ============================================================
# Student exercise: 3D scatter plot of study hours vs sleep hours vs exam scores
#
# Instruction:
# Create an interactive 3D scatter plot using Plotly to visualize relationships between study_hours, 
# sleep_hours, and exam_score, colored by major.
# Plotly allows you to interact with the plot - zoom, pan, and rotate in 3D space.
# Be sure to add a title and label the axes.
# ============================================================
# Step: import plotly.express as px
import plotly.express as px

# Step: prepare the data by selecting relevant columns and removing missing values
subset = df[['study_hours', 'sleep_hours', 'exam_score', 'major']].copy()
subset = subset.dropna()

# Step: create the interactive 3D scatter plot with title and axis labels
fig = px.scatter_3d(subset, 
                    x='study_hours', 
                    y='sleep_hours', 
                    z='exam_score',
                    color='major',
                    labels={'study_hours': 'Study Hours (per week)',
                           'sleep_hours': 'Sleep Hours (per night)',
                           'exam_score': 'Exam Score'},
                    title='3D Scatter: Study Hours vs Sleep Hours vs Exam Scores by Major')

# Step: adjust marker size for better visibility
fig.update_traces(marker=dict(size=1))

# Step: display the interactive plot
fig.show()
```
