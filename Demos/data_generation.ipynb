```python
import numpy as np
import pandas as pd

np.random.seed(123)

# Number of individuals
n = 700

# Categories
activity_level = ["Sedentary", "Lightly Active", "Moderately Active", "Very Active"]
age_group = ["18-25", "26-35", "36-45", "46-55", "56+"]

# Base numeric data
weekly_workouts = np.random.poisson(3, n).clip(0, 10)               # sessions/week
avg_steps = (np.random.normal(7000, 2500, n) + weekly_workouts*500).clip(1000, 20000)
sleep_hours = np.random.normal(7.1, 0.9, n).clip(4, 10)             # hours/night
water_intake = np.random.normal(1.8, 0.7, n).clip(0.3, 4.5)         # liters/day

# Derived relationships
# - More workouts and more sleep → lower resting HR
resting_hr = (
    75
    - 1.5 * weekly_workouts
    - 0.8 * (sleep_hours - 7)
    + np.random.normal(0, 4, n)
).clip(45, 110)

# VO2 estimate based on steps + workouts
vo2_estimate = (
    30
    + 0.0007 * avg_steps
    + 0.8 * weekly_workouts
    + np.random.normal(0, 3, n)
).clip(15, 60)

# Fitness level category (based on VO2 estimate)
bins = [0, 25, 35, 45, 60]
labels = ["Poor", "Fair", "Good", "Excellent"]
fitness_level = pd.cut(vo2_estimate, bins=bins, labels=labels, include_lowest=True)

# Assemble final DataFrame
df = pd.DataFrame({
    "activity_level": np.random.choice(activity_level, n),
    "age_group": np.random.choice(age_group, n),
    "weekly_workouts": weekly_workouts,
    "avg_daily_steps": avg_steps,
    "sleep_hours": sleep_hours,
    "water_liters": water_intake,
    "resting_heart_rate": resting_hr,
    "vo2_estimate": vo2_estimate,
    "fitness_level": fitness_level,
})

# Save to listings_synthetic.csv (replacing the old synthetic dataset)
df.to_csv('synthetic_data.csv', index=False)
print("Dataset saved to synthetic_data.csv")

df.head()
```
