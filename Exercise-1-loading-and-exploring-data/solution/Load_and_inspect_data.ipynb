```python
# Display information about the DataFrame (data types, memory usage, etc.)
df.info()
```

### Get statistical summary of the DataFrame
Use the `describe()` method to display statistical summary of the numerical columns in the DataFrame.

```python
# Display statistical summary of the DataFrame (e.g. mean, std, min, max, etc.)
df.describe()
```

Use the `dtypes` attribute to display the data type of each column in the DataFrame.

## Working with Polars

Now let's repeat the same operations using **Polars**, a modern DataFrame library designed for speed and efficiency. While Polars shares many concepts with pandas, you'll notice some differences in syntax and method names that reflect its design philosophy.

```python
# Import polars
import polars as pl
# Path handles file paths consistently across operating systems
from pathlib import Path

# Set the path to the csv file
# (synthetic dataset in the Exercises folder)
data_path = Path('../starter/synthetic_data.csv')

# Load the data into a polars DataFrame using pl.read_csv()
df_pl = pl.read_csv(data_path)
```

### Inspect the first few rows
Use the `head()` method to display the first 5 rows of the DataFrame. This gives you a quick overview of the data structure and values.


```python
# Display the first 5 rows of the DataFrame
# The head() method works similarly to pandas, showing the first 5 rows by default
df_pl.head()
```

### Get information about the DataFrame
Use the `shape`, `schema`, and `estimated_size()` methods to display information about the DataFrame including the number of rows, columns, data types, and memory usage.


```python
# Display information about the DataFrame (shape, schema, and estimated size)
# Polars doesn't have a single info() method like pandas, so we use multiple methods:
# - shape: returns (rows, columns) as a tuple
# - schema: returns a dictionary-like object showing column names and data types
# - estimated_size(): returns the estimated memory usage in the specified unit
print(f"Shape: {df_pl.shape}")
print(f"\nSchema:")
print(df_pl.schema)
print(f"\nEstimated size: {df_pl.estimated_size('mb'):.2f} MB")
```

### Get statistical summary of the DataFrame
Use the `describe()` method to display statistical summary of the numerical columns in the DataFrame.


```python
# Display statistical summary of the DataFrame (i.e. summary statistics for numerical columns)
df_pl.describe()
```

### Check the data types of each column
Use the `dtypes` attribute or `schema` property to display the data type of each column in the DataFrame.


```python
# Display the data types of each column
# In Polars, you can use either dtypes (returns a list) or schema (returns a dictionary-like object with more details)
df_pl.dtypes
```
